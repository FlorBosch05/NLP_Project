# MVP options and how to compare them

This map covers the choices explicitly named in the project brief. “Or alternatives”, “or something similar” and “e.g.” do not require a specific additional product. The Extra Mile items are separate extensions and are not counted as MVP alternatives. The existing implementation is V2; these comparison notebooks do not replace it.

| Decision in the brief | Named options | Existing MVP | Where to compare |
|---|---|---|---|
| Corpus | Curate or scrape 30–50 recipes | 30 synthetic examples; times and labels unverified | Notebook 12; source audit needed |
| Chunking | Sentence/token, semantic, hierarchical separating ingredients and instructions | Section and whole-recipe chunks | Notebook 09 tests sentence, token, semantic, sections, parent-child and whole recipe |
| Local vector store | Chroma, Qdrant, FAISS | Chroma | Notebook 10 tests all three with identical vectors and queries |
| Model/inference | Groq free tier, example `llama-3.1-8b-instant` | Groq `openai/gpt-oss-20b` | Notebook 12; the example has been withdrawn on free/developer tiers |
| Orchestration | LlamaIndex and LangGraph, or alternatives | Both | Notebook 12; no specific alternatives named in brief |
| MCP | `langchain-mcp-adapters` or `FastMCP`; tool or pantry resource | FastMCP tools, direct MCP client | Notebook 11 exercises both primitives and both connection approaches |
| Evaluation | Context precision, context recall, faithfulness with Ragas; 15–20 queries | 20 test cases, partial Ragas scores | Notebook 12 and existing evaluation notebooks |
| UI | Streamlit or similar | Streamlit | Notebook 12; no other UI named in the brief |

## Fair comparisons

Keep the corpus, embedding model, queries and relevance labels fixed. Compare quality and latency separately and inspect failures. The snippets here use a small synthetic dataset and simple relevance proxies, not independently verified ground truth. The vector-store notebook uses in-memory indexes for all three engines; it does not compare disk persistence or production capacity. No comparison should be represented as a Ragas benchmark unless Ragas was actually run.

To prepare optional libraries, from NLPproject run `uv pip install -r v2/comparisons_v1/requirements-comparisons.txt`. Use the same `.venv` kernel as the MVP. The experiments do not call Groq.
